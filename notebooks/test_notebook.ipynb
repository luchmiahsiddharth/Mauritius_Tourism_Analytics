```python

df= pd.read_excel('C:\\Users\\sidlu\\Mauritius_Tourism_Analytics\\data\\raw\\HS_Tourism_Yr25_220526.xls', sheet_name="Table 2 ", header=None)

df.columns=df.iloc[3]

df.columns.name=None

df=df.drop(range(0,4)).reset_index(drop=True)

df = df.melt(
    id_vars='Country of disembarkation',   # column(s) to keep as-is
    var_name='Year',                        # name for the new "year" column
    value_name='Arrivals'                   # name for the new "value" column
)
df=df.sort_values(["Country of disembarkation","Year"]).reset_index(drop=True)

df.head(50)
```

```python
rows_to_remove = ['All countries', 'Others & not stated', '\"Source: Administrative data, Passport and Immigration Office\"', '2 Disembarkation may either be the country of final destination or the transit country', '1 Excludes inter-island traffic between the islands of Mauritius and Rodrigues', '3 Special Administrative Region of China', 'n.ap: not applicable']

df=df[~df['Country of disembarkation'].isin(rows_to_remove)]

df = df[df['Country of disembarkation'].notna()]

df = df[~df['Country of disembarkation'].str.startswith('Source:', na=False)]
```

```python
df.to_csv("C:\\Users\\sidlu\\Mauritius_Tourism_Analytics\\data\\processed\\HS_Tourism_Clean.CSV", index=False)
```
